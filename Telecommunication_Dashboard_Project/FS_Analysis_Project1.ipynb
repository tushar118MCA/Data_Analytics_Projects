```python
import pandas as pd
import os
import re
from sqlalchemy import create_engine

# -----------------------------
# STEP 1: User Input
# -----------------------------
user_sector = input("Which Sector U want : ").strip()
user_igroup = input("Which Igroup U want: ").strip()

# Creating a clean database table name by removing special characters and spaces
# E.g., 'Energy and Power Generation' becomes 'pnl_energy_power_generation'
clean_sector = re.sub(r'[^a-zA-Z0-9_]', '', user_sector.lower().replace(' ', '_'))
clean_igroup = re.sub(r'[^a-zA-Z0-9_]', '', user_igroup.lower().replace(' ', '_'))
target_table_name = f"pnl_{clean_sector}_{clean_igroup}"

username = "postgres"
password = "125890"
host = "localhost"
port = "5432"
database = "FS_Analysis"

# Creating a SQLAlchemy Engine
engine = create_engine(f"postgresql+psycopg2://{username}:{password}@{host}:{port}/{database}")

# -----------------------------
# STEP 2: Data Fetching Clause (your original SQL)
# -----------------------------
conn.rollback() # Rollback of your original connection

try:
    print(f"\n⏳ Sector: '{user_sector}' And Igroup: '{user_igroup}' Data Loading from Database...")
    
    whole_data_sql = """
    SELECT 
        ROW_NUMBER() OVER (ORDER BY c."Company Symbol" ASC, pl."Financial Year" DESC) AS row_num,
        
        -- १. Company 
        c."Company Symbol",
        c."Company Name",
        c."Sector" AS company_sector,
        c."Industry",
        c."Igroup",
        c."MarketCap",
        
        -- ②. Financial Year & Quarter
        pl."Financial Year" AS financial_year,
        pl."Quarter" AS pl_quarter,
        
        -- ③. Taxonomy to Element Id
        (string_to_array(pl."Taxonomy_Id", '.'))[cardinality(string_to_array(pl."Taxonomy_Id", '.'))] AS element_id,
        pl."Taxonomy_Id" AS pl_taxonomy_id,
        
        -- ④. P&L YTD 
        pl."Element Name" AS pl_element_name,
        pl."Value" AS pl_value,
        CASE WHEN pl."Decimal" = '7' THEN '0' ELSE pl."Decimal" END AS pl_decimal,
        
        pl."Unit" AS pl_unit

    FROM public."Company" c
    LEFT JOIN public."P&L YTD" pl ON c."Company Symbol" = pl."Company Symbol"
    WHERE c."Sector" = %s AND c."Igroup" = %s;
    """
    
    # Reading data in chunks
    chunks = pd.read_sql_query(whole_data_sql, conn, params=(user_sector, user_igroup), chunksize=10000)
    
    df_list = []
    for chunk in chunks:
        df_list.append(chunk)
        
    if len(df_list) > 0:
        df = pd.concat(df_list, ignore_index=True)
    else:
        df = pd.DataFrame() 
    
    if df.empty:
        print(f" Data Not Found! Please check spelling of Sector or Igroup.")
    else:
        print(f" Data Fetched Successfully! Total Rows: {len(df)}")
        print(f" Creating and Loading table '{target_table_name}' into PostgreSQL...")
        
        # Pushing data directly to SQL instead of Excel (using `if_exists='replace'` will overwrite the existing table if it exists)
        df.to_sql(
            name=target_table_name, 
            con=engine, 
            schema="public", 
            if_exists="replace", 
            index=False
        )
        
        print(f"Successfully'{target_table_name}' name table created")

except Exception as e:
    print(f" Error Occurs: {e}")
```

```python
# -------------------------------------------------------------------
# Creating an Excel file named after the database table (target_table_name)
# -------------------------------------------------------------------
try:
    if not df.empty:
       # Setting the Excel file name to match the database table name (e.g., pnl_energy_oil.xlsx)
        exact_excel_file_path = f"{folder_path}/{target_table_name}.xlsx"
        
        # Create the folder if it does not exist
        if not os.path.exists(folder_path):
            os.makedirs(folder_path)
            
        print(f"\n Generating Excel file at: {exact_excel_file_path}...")
        
        df.to_excel(exact_excel_file_path, index=False, engine='openpyxl')
        
        print(f"Excel file saved successfully!")
        print(f"File name and path: {exact_excel_file_path}")
    else:
        print("\nCould not create the Excel file because the data frame is empty..")

except Exception as e:
    print(f"An error occurred while creating the Excel file.: {e}")
```
